```python
from ioMicro import read_im
import numpy as np
```

```python
# open napari
viewer = napari.Viewer()
```

```python
# read image and load into napari
img = read_im(r'Z:\Zane_20CRE\7_17_2024__T7_20CRE\HGFP\Conv_zscan__150.zarr')
viewer.add_image(img)
```

```python
# load the drift dictionary for a specific FOV
drifts = np.load(r'Z:\Zane_20CRE\7_17_2024__T7_20CRE\drifts_polyA\Conv_zscan__150_drift.pkl', allow_pickle=True)
```

```python
# load thresholds for T7 data
import numpy as np
h_ths = np.load('final_parms_ths.npz')['final_parms']
h_ths = h_ths.astype(int)
```

```python
# load spots for a specific enhancer, correct for drift, and add spots (filtered by brightness) to the napari viewer
CRE = np.load(r'Z:\Zane_20CRE\7_17_2024__T7_20CRE\analysis\Conv_zscan__150--H1_fits_icol2.npz')['Xh']
CRE_drift = drifts['H1'][0]
CRE[:,:3]-=CRE_drift
h = CRE[:,7]
keep = np.where((h>150) & (h<5000))
CRE_ = CRE[keep]
viewer.add_points(CRE_[:,:3], size=8, face_color='r', blending='additive')
```

```python
# load spots for a specific T7 transcript, correct for drift, and add spots (filtered by brightness) to the napari viewer
T7 = np.load(r'Z:\Zane_20CRE\7_17_2024__T7_20CRE\analysis\Conv_zscan__150--D1_fits_icol2.npz')['Xh']
T7_drift = drifts['D1'][0]
T7[:,:3]-=T7_drift
h = T7[:,7]
keep = np.where(h > 950)
T7_ = T7[keep]
viewer.add_points(T7_[:,:3], size=8, face_color='green', blending='additive')
```
