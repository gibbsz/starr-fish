```python
from plastid import GenomicSegment, SegmentChain, Transcript
```

```python
# CRE001 transcript, containing 100 bp of 5' UTR and 150 bp of 3' UTR
# 714bp coding region from bases 945-1659
CRE001 = Transcript(GenomicSegment("SFv6",0,570,"+"),ID="CRE001",cds_genome_start=0,cds_genome_end=570)
CRE002 = Transcript(GenomicSegment("SFv6",0,570,"+"),ID="CRE002",cds_genome_start=0,cds_genome_end=570)
```

```python
# make list of CREs 
CRE_list = ['CRE' + str(f'{i:03}') for i in range(0,300)]

# iterate through CRE list to generate annotations for each CRE
# look up plastid package for more info
for i in CRE_list:
    i = Transcript(GenomicSegment(i,0,380,"+"),gene_id=i,ID=i,cds_genome_start=0,cds_genome_end=380)
    fout = open("CRE300_ann.gtf","a")
    fout.write(i.as_gtf())
    fout.close()
```
