```python
from Bio import SeqIO

# Define the path to your mouse genome FASTA file
fasta_file = r'C:\Users\zgibbs\mm10.fa.gz'

# Example: Define a list of coordinates (chromosome, start, end) to extract
coordinates = [
    ('chr1', 1000, 2000),  # Extract sequence from chr1 from position 1000 to 2000
    ('chr2', 500, 1500),   # Extract sequence from chr2 from position 500 to 1500
]

# Create an empty dictionary to store extracted sequences
extracted_sequences = {}

# Read the FASTA file and extract the specific regions based on coordinates
for record in SeqIO.parse(fasta_file, "fasta"):
    for chrom, start, end in coordinates:
        if record.id == chrom:
            # Extract the sequence from start to end (1-based indexing in Genomics)
            extracted_sequences[chrom] = record.seq[start-1:end]

# Print the extracted sequences
for chrom, sequence in extracted_sequences.items():
    print(f">{chrom}_extracted")
    print(sequence)
```

```python
from Bio import SeqIO
from Bio.SeqRecord import SeqRecord

# Specify the path to your input FASTA file and the output FASTA file
input_fasta_file = r'C:\Users\zgibbs\STARR-FISH_positive_controls\pos_ctrls.fa'
output_fasta_file = r'C:\Users\zgibbs\STARR-FISH_positive_controls\pos_ctrls_central_190bp.fa'

# Create a list to store the new SeqRecord objects
new_records = []

# Loop over each sequence in the input FASTA file
for record in SeqIO.parse(input_fasta_file, "fasta"):
    seq_length = len(record.seq)
    
    # Ensure the sequence is long enough to extract 190 bp
    if seq_length >= 190:
        # Calculate the start and end positions for the central 190 bp
        start = (seq_length - 190) // 2
        end = start + 190
        
        # Extract the central 190 bp
        central_190_bp = record.seq[start:end]
        
        # Create a new SeqRecord object for the extracted sequence
        new_record = SeqRecord(central_190_bp, id=record.id, description="central_190bp")
        
        # Add the new SeqRecord to the list
        new_records.append(new_record)
    else:
        print(f"Sequence {record.id} is too short to extract 190 bp.")

# Write the extracted sequences to the new FASTA file
with open(output_fasta_file, "w") as output_handle:
    SeqIO.write(new_records, output_handle, "fasta")

print(f"Extracted sequences have been written to {output_fasta_file}.")
```
